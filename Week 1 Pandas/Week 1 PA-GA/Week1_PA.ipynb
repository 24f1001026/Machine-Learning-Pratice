##### Q5 : Convert the values of 'Duration' into seconds. Enter the average duration (in seconds) of a flight. Enter your answer correct to two decimal places.


````python
# data['Duration']

def convert_to_seconds(duration):
    parts = duration.split()
    
    hours = 0
    minutes = 0
    
    for part in parts:
        if 'h' in part:
            hours = int(part.replace('h', ''))
        elif 'm' in part:
            minutes = int(part.replace('m', ''))
    
    seconds = hours * 3600 + minutes * 60
    
    return seconds


data['Duration']

# data['Duration'].apply(convert_to_seconds)
data['Duration'].apply(convert_to_seconds).mean().round(2)
````

## 5. Frequency Analysis – `value_counts()`

**Concept:** Counts how many times each unique value appears, sorted high → low by default.

```python
data['Month'].value_counts()                    # counts per month
data['Month'].value_counts(ascending=True)      # low -> high
data['Month'].value_counts(normalize=True)      # proportions instead of counts
data['Month'].value_counts().idxmax()           # the VALUE (label) with highest count
data['Month'].value_counts().max()              # the highest COUNT itself
```

**`max()` vs `idxmax()`:**
- `max()` → the largest number (e.g., 2500 flights)
- `idxmax()` → the label of that largest number (e.g., month 5)

## 17. Duration Processing

**Concept:** Strings like `'2h 50m'`, `'3h'`, `'45m'` → convert to numeric seconds.
`seconds = hours × 3600 + minutes × 60`

```python
def convert_to_seconds(duration):
    hours = 0
    minutes = 0
    for part in duration.split():
        if 'h' in part:
            hours = int(part.replace('h', ''))
        elif 'm' in part:
            minutes = int(part.replace('m', ''))
    return hours * 3600 + minutes * 60

data['Duration'] = data['Duration'].apply(convert_to_seconds)
data['Duration'].mean().round(2)
```

> Initialising `hours = 0` and `minutes = 0` handles entries that have only one part (e.g., `'3h'`).

## 21. Common Mistakes to Avoid

1. Using `and` / `or` instead of `&` / `|` on Series.
2. Forgetting parentheses around each condition.
3. Forgetting to assign back after `replace()` or `apply()`.
4. Writing `apply(func())` instead of `apply(func)`.
5. Confusing `max()` (the count) with `idxmax()` (the label).
6. Using `[ ]` instead of `[[ ]]` when selecting multiple columns.
7. Re-running a conversion cell twice (e.g., `get_hour` on an already converted column causes an error).
